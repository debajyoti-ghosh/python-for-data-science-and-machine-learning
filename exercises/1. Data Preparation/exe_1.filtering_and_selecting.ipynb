# 6. Challenge Exercise

i. Create a MultiIndex DataFrame with Region → City → Year. Fill it with random sales data.

ii. Select all rows for Region='Europe', City='Amsterdam'.

iii. Slice the DataFrame to show only data for years 2021–2023.

iv. Use Boolean indexing to display sales greater than 5000.

v. Add another column with random profit data.

vi. Use Boolean indexing to display profit greater than 500.

# Solution

i. Create a MultiIndex DataFrame with Region → City → Year. Fill it with random sales data.


Step 1: Define data

```python
years = list(range(2020, 2026))
regions = ['Asia', 'Europe','North America','South America']
cities_per_region = [
    ["Tokyo", "Beijing", "Mumbai", "Bangkok"],
    ["Amsterdam", "Berlin", "Madrid", "Rome"],
    ["New York", "Los Angeles", "Chicago", "Houston"],
    ["São Paulo", "Buenos Aires", "Lima", "Bogotá"]
]
```

Step 2: Flatten region-city combinations

```python
region_city_pairs = [(region, city)
                     for region, cities in zip(regions, cities_per_region)
                     for city in cities]

region_city_pairs
```

Step 3: Create MultiIndex with Region → City → Year

```python
index = pd.MultiIndex.from_product(
    [region_city_pairs, years],
    names=['region_city', 'year']
)

# Split the tuple into two separate levels: region and city
index = pd.MultiIndex.from_tuples(
    [(region, city, year) for (region, city), year in index],
    names=['region', 'city', 'year']
)
index
```

Step 4: Create DataFrame with random sales

```python
sales_data = pd.DataFrame(
    np.random.randint(1000, 8000, size=(len(index), 1)),
    index=index,
    columns=['sales']
)

print(sales_data.head(36))
# sales_data
```

ii. Select all rows for Region='Europe', City='Amsterdam'.

```python
sales_data = sales_data.sort_index() # Sort the MultiIndex for easier access, otherwise will have PerformanceWarning: indexing past lexsort depth may impact performance

sales_data.loc[('Europe','Amsterdam'), :]
```

iii. Slice the DataFrame to show only data for years 2021–2023.

```python
idx=pd.IndexSlice
sales_data_2021_23 = sales_data.loc[idx[:, :, 2021:2023], : ]
sales_data_2021_23
```

iv. Use Boolean indexing to display sales greater than 5000.

```python
sales_data[sales_data > 5000]
```

```python
sales_data[sales_data["sales"] > 6000]
```

```python
sales_data.loc[sales_data['sales'] > 5000, 'sales']
```

```python
sales_data[(sales_data['sales'] > 5000) & (sales_data.index.get_level_values('year') >= 2022) & (sales_data.index.get_level_values('region') == 'Asia')]
```
